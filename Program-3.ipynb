```python
import cv2
import numpy as np
from scipy.signal import convolve2d
from scipy.ndimage import convolve
import matplotlib.pyplot as plt
```

```python
# Read the image
image = cv2.imread('/content/b0755-16425315464091-1920.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
# Define kernels
kernels = {
    'Identity': np.array([[0, 0, 0], [0, 1, 0], [0, 0, 0]]),
    'Edge Detection': np.array([[1, 0, -1], [0, 0, 0], [-1, 0, 1]]),
    'Sharpen': np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]]),
    'Box Blur': np.array([[1, 1, 1], [1, 1, 1], [1, 1, 1]]) / 9,
    'Gaussian Blur': np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]]) / 16
}
```

```python
# Apply kernels using filter2D
for kernel_name, kernel in kernels.items():
    filtered_image = cv2.filter2D(image, -1, kernel)
    plt.imshow(filtered_image, cmap='gray')
    plt.title(f'{kernel_name} Applied')
    plt.axis('off')
    plt.show()
```

```python
# Apply kernels using convolve function in SciPy
for kernel_name, kernel in kernels.items():
    convolved_image = convolve(image, kernel, mode='nearest')
    plt.imshow(convolved_image, cmap='gray')
    plt.title(f'{kernel_name} Applied (SciPy)')
    plt.axis('off')
    plt.show()
```
